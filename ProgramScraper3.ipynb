```python
def get_link(a_tag):
  """
  Takes an course's <a> tag that contains an "onclick" parameter, and finds the relevant courses' link.
  Returns a list of the course's name and appropriate link.

  When implemented, This function will pass the link to the url_scraper(url) function to grab any prerequisites or GER credits.
  """
  # just to make the get_all_tags() code less repetitive
  # if a_tag.name == "a" or a_tag.find("a"):
  if a_tag.name == "a":
    link_info = a_tag["onclick"]
  elif a_tag.find("a"):
    link_info = a_tag.a["onclick"]

  course = a_tag.text.replace("\xa0", " ").strip()

  if link_info.startswith("showCourse"):
    catoid = link_info.split("'")[1]
    coid = link_info.split("'")[3]
    link = overhead + "preview_course.php?catoid={}&coid={}&print".format(catoid, coid)
  elif link_info.startswith("acalogPopup"):
    link = overhead + link_info.split("'")[1]

  # print(link)
  # course_info = url_scraper(link)

  # return [course, course_info]
  return [course, link]


```

```python
def get_all_tags(list_element):
  """
  Takes a list element tag <li>. Parses through the tag and gets all relevant course information.
  Program Descriptions (a div), any tags like like <p> tags, <strong> tags, <a> tags, anything that
  may have importance for determining a courses information is grabbed and return as a list.

  This calls the get_link(tag) function to retrieve links for each course in the program.
  """
  # sometimes, a NavigableString is passed.
  list_text = list_element.text.replace("\xa0", " ").strip() if not isinstance(list_element, NavigableString) else str(list_element).replace("\xa0", " ").strip()
  if isinstance(list_element, NavigableString):
    if list_text != "":
      return [list_text]
    else:
      return []

  # Sometimes, a <li> has no lists. This can have a class in it, or more likely textual information
  elif not list_element.has_attr("class"):
    # if not isinstance(list_element, NavigableString):
    # print("WHY ART/COM: ", list_element)
    if list_element.find("a") != None:
      course_info = []
      for tag in list_element:
        # print("CLASSLESS TAG: ", tag)
        tag_text = tag.text.replace("\xa0", " ").strip() if not isinstance(tag, NavigableString) else str(tag).replace("\xa0", " ").strip()
        if not isinstance(tag, NavigableString):
          # print("TAG: ", tag)
          if tag.find("a") or tag.name == "a":
            course_info.append(get_link(tag))

          else:
            if tag_text != "":
              course_info.append(tag_text)
        # if it is a NavigableString, it can't use .text, so format is slightly different
        else:
          if tag_text != "":
            course_info.append(tag_text)
      return course_info
    else:
      # print("IG NO CLASSES?")
      if list_text != "":
        return [list_text]

  # this concerns <li>'s that specifically have courses in them. These usually have links
  # sometimes, there is textual information like "course1 or course2".
  elif "acalog-course" in list_element["class"]:
    # more than 1 course on a line
    # if len(list_element.span) > 1:
      course_info = []
      for tag in list_element.span:
        # print("TAG: ", tag)
        tag_text = tag.text.replace("\xa0", " ").strip() if not isinstance(tag, NavigableString) else str(tag).replace("\xa0", " ").strip()
        if not isinstance(tag, NavigableString):
          if tag.name == "a" or tag.find("a"):
            course_info.append(get_link(tag))

          # sometimes, a program decides to put info as a paragraph (ex: WGS)
          # othertimes, a program decides to put info as text (ex: BIOMDS), usually OR or AND in the form of <strong>
          elif tag.name in ("p", "strong", "span"):
            # print(tag)
            # apparently, some groups put a course in a strong tag with an or
            # make sure to filter out the a (looking at you ANT)
            if tag.name == "strong" and tag.find("a") != None:
              # link_info = tag.find("a")["onclick"]
              # link = overhead + link_info.split("'")[1]
              # course = tag.text.replace("\xa0", " ").strip()
              course_info.append(get_link(tag))

            elif tag_text != "":
              course_info.append(tag_text)
            # course_info.append(tag.text)

          elif re.search(r'<|>', str(tag)) == None and tag_text != "":
              course_info.append(tag_text)
            # print(tag)
            # course_info.append(tag)
          # thanks Comp Sci
          # elif tag.name == "span":
          #     print(tag.text)
        # NavigableString has different format - no .text
        else:
          if tag_text != "":
            course_info.append(tag_text)
      return course_info

  # adhoc corresponds to important info (NOT a course), but not necessarily found in a <li> tag (ex: BIOMDS notes)
  # adhoc-list-item is the same thing, but specifically as a <li> tag
  # these can also harness links, usually if they follow text first (looking at you Art history)
  elif "acalog-adhoc" in list_element["class"] or "acalog-adhoc-list-item" in list_element["class"]:
      # print("fuck me", list_element.name, list_element)
      if list_element.find("strong") and list_element.strong.find("a") != None:
        course_info = []
        for tag in list_element.strong:
          tag_text = tag.text.replace("\xa0", " ").strip() if not isinstance(tag, NavigableString) else str(tag).replace("\xa0", " ").strip()
          # print("does this ever run? :", tag)

          if not isinstance(tag, NavigableString):
            if tag.name == "a" or tag.find("a"):
              course_info.append(get_link(tag))
            else:
              if tag_text != "":
                course_info.append(tag_text)
          else:
            if tag_text != "":
              course_info.append(tag_text)
        return course_info

      elif list_element.find("a") != None:
        course_info = []
        for tag in list_element.children:
          # print("ADHOC TAG: ", tag)
          tag_text = tag.text.replace("\xa0", " ").strip() if not isinstance(tag, NavigableString) else str(tag).replace("\xa0", " ").strip()
          if not isinstance(tag, NavigableString):
            if tag.find("a") or tag.name == "a":
              course_info.append(get_link(tag))

            else:
              if tag_text != "":
                course_info.append(tag_text)
          else:
            if tag_text != "":
              course_info.append(tag_text)
        return course_info

      else:
        if list_text != "":
          return [list_text]
  # if this code somehow runs, god help you. Acalog's formatting was pain enough.
  else:
    pass
    print("WHAT: ", list_element)
```

```python
def program_scrape(program_idx):
  """
  This function takes in an index of the program_results list, which contains a program name and url.
  It soups this url, and (utilizing get_all_tags(tag)) returns all information on the page in a list of lists.
  It itself grabs the description and header tags, but any lists are parsed separately in the get_all_tags(tag) function.

  This is run on every single program on the list to eventualyl return a dataframe containing all the information on
  each program on the Furman 2024-2025 Undergraduate Academic Catalog (https://catalog.furman.edu).
  """
  program_lst = []
  p = program_results[program_idx]
  p_content = requests.get(p["url"])
  p_soup = BeautifulSoup(p_content.text, "html.parser")
  div_layer = p_soup.find(attrs={"class":"block_content"})
  # print(p["name"], p["url"])
  groupings = div_layer.find_all("tr")
  program_lst.append([p["name"], p["url"]])

  for item in groupings:
    if not isinstance(item, str):
      pDesc = item.find("div", class_="program_description")
      pDesc_text = pDesc.text.replace("\xa0", " ") if pDesc else "\n"
      if pDesc:
        # programs_df['Description'] = pDesc.text
        if pDesc_text != "\n":
          program_lst.append([pDesc_text])
        else:
          program_lst.append([])
      # If its the actual courses info
      else:
        cores = item.find_all("div", "acalog-core")

        for core in cores:
          header_lst = []
          # print("Core tag: ", core)
          # print("\n")
          for child in core.children:
            # print(len(child))
            # print("CHILD: ", child)
            # if header or li, must have more than one element
            if re.search(header_pattern, str(child)):
                header_type = re.search(header_pattern, str(child))[0]
                full_header = header_type + " " + child.text
                # print("HEADER CHILD: ", child)
                # print("HEADER TEXT", full_header)
                header_lst.append(full_header)

            elif re.search(ul_pattern, str(child)):
                # print("UL CHILD: ", child)
                for tags in child.children:
                  # courses = get_all_links(tag)
                  courses = get_all_tags(tags)
                  header_lst.append(courses)
                  # print("C: ", c)
                # print("")
            elif re.search(r'(?<=<)p(?=>)', str(child)):
              if child.text.replace("\xa0", " ").strip() != "":
              # print("P CHILD: ", child)
                header_lst.append(child.text.replace("\xa0", " ").strip())

            # programs_df['Info'] += (header_dict)
          program_lst.append(header_lst)
  return program_lst
```
